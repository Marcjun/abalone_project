En un principio se crearon las 3 branches (ramas), partiendo desde el main, con el objetivo de modificar la misma parte (parametros del train_model), para luego hacer una PR de las 3 hacia el main, con el fin de ver como entramos en conflicto

![Captura de pantalla 2026-10-25 204450.png](<attachment:Captura de pantalla 2026-10-25 204450.png>)


En la imagen se puede apreciar una ilustracion de como figuraba una de las ramas luego de aprobar la PR de un individuo

Como se puede ver, una vez aprobada la primer PR no deja aceptar la de los demas individuos. 

Por lo que seguimos con el objetivo de solucionar este problema a mano como se nos indico.

![Captura de pantalla 2026-10-25 205148.png](<attachment:Captura de pantalla 2026-10-25 205148.png>)

Aqui se puede ver como es que aparecen los cambios de la PR aceptada y con la PR que no se pudo aceptar, las opciones para poder tratar el problema son las que se nos muestran arriba, Accept Current change | Accept Incomign change | Accept both changes | compare changes

Dependiendo del tipo de error que se aprecie usamos alguna de estas opciones o intervenimos manualmente con el codigo.

Por ejemplo, en la imagen se ve que el error es el cambio de parametros, entonces el punto medio elegido fue hacer una "mezcla" de ambos limites de los parametros seleccionados por LOS usuarios, es decir, se agrando el intervalo con el objetivo de que ambas opciones se han consideradas en el resultado final

Por otro lado, algunos cambios solo "se aceptaron ambos" como en el archivo "experiments_log.csv" se realizo ya que el objetivo era mantener las metricas y hiperparametros usados en el entrenamiento. Se realizo una intervencion manual alli, al ver que ambos resultados tenian el mismo ID. 
Lo mismo ocurre con el "metrics_v2.json", se aceptaron ambos, lo cual incurrio en un error que luego se corrigio
